```python
with open('sample.txt','w') as file:
    file.write("hello world\n")
    file.write("Created by pratik\n")
    file.write("Hello pratik\n")
```

```python
# overwriting theb entire file 

with open('sample.txt','r') as file:
    
    for line in file:
        print(line.strip())
```

```python
## without over writing

with open('sample.txt','a') as file:
    file.write("over write is taking place\n")
```

```python
## write multiple lines 

lines=['first line\n','second line\n']

with open('sample.txt','a') as file:
    file.writelines(lines) 
```

```python
## write a binary file
data=b'\x00 \x01 \x02 \x03'
with open('sample.bin','wb')as file:
    file.write(data)
```

```python
## read a binary file

with open('sample.bin','rb') as file:
    text=file.read()
    print(text)
```

```python
## move the cursor to first

with open('sample.txt','r') as file:
    file.seek(0)
    content=file.read()
    print(content)
```
